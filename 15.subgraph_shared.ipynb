```python
from langchain_mistralai import ChatMistralAI
from typing import TypedDict
from langgraph.graph import START, END, StateGraph

from dotenv import load_dotenv
load_dotenv()
```

```python
llm = ChatMistralAI()
```

```python
class ParentState(TypedDict):
    question: str
    english_answer: str
    hinglish_answer: str
```

```python
def translate_in_hinglish(state: ParentState):
    prompt = f"Translate the followinr engish text into hinglish and do not add any extra imformation text: {state['english_answer']}"
    answer = llm.invoke(prompt).content
    return {"hinglish_answer": answer}
```

```python
sub_builder = StateGraph(ParentState)

sub_builder.add_node("translate_in_hinglish", translate_in_hinglish)

sub_builder.add_edge(START, "translate_in_hinglish")
sub_builder.add_edge("translate_in_hinglish", END)

sub_graph = sub_builder.compile()
```

```python
def generate_answer(state: ParentState):
    prompt = state["question"]
    answer = llm.invoke(prompt).content
    return{"english_answer": answer}
```

```python
parent_builder = StateGraph(ParentState)

parent_builder.add_node("generate_answer", generate_answer)
parent_builder.add_node("sub_graph", sub_graph)

parent_builder.add_edge(START, "generate_answer")
parent_builder.add_edge("generate_answer", "sub_graph")
parent_builder.add_edge("sub_graph",END)

parent_graph = parent_builder.compile()

result = parent_graph.invoke({
    "question": "where is manali"
})
print(result["hinglish_answer"])
```
